```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
```

```python
class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3)
        self.fc1 = nn.Linear(16*26*26, 10)
    
    def forward(self, x):
        x = self.conv1(x)
        x = x.view(x.size(0), -1)
        x = self.fc1(x)
        return x
```

```python
model = SimpleNet()
torch.onnx.export(model, torch.randn(1, 3, 28, 28), "simplenet.onnx")
```
